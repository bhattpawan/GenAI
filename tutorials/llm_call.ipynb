```python
from google import genai
from pprint import pprint
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
model = "gemini-3.5-flash-lite"
user_input = input("Enter your query:")
API_KEY = os.getenv("GOOGLE_API_KEY")

client = genai.Client(api_key=API_KEY)
```

```python
def get_api_response(input):
    interaction = client.interactions.create(model=model, input=input)
    return interaction.output_text
```

```python
def stream_api_response(input):
    stream = client.interactions.create(model=model, input=input, stream=True)
    for event in stream:
        if event.event_type == "step.delta":
            if event.delta.type=="text":
                pprint(event.delta.text)
```

```python
stream = client.interactions.create(model=model, input=user_input, stream=True)
for event in stream:
    pprint(event)
```

```python
try:
    # response = get_api_response(user_input)
    # print(response)
    stream_api_response(user_input)
except Exception as e:
    print(f"An error occurred: {e}")
```
